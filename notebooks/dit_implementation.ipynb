```python
import torch
from diffusion_hub.models import get_model
```

```python
from dataclasses import dataclass, asdict

@dataclass
class Config:
    res: int = 32
    n_channels: int = 3
    patch_size: int = 4
    dim: int = 384
    n_blocks: int = 8
    n_heads: int = 6
    time_emb_dim: int = 256
    time_emb_scale: float = 1000.0
    n_classes = None
    attn_dropout_p: float = 0.0
    bias: bool = True

cfg = Config()

model = get_model("dit", **asdict(cfg)).cuda()
```

```python
x = torch.randn(16, 3, 32, 32).cuda()
t = torch.rand(16).cuda()

out = model(x, t)

print(out.shape)
```
